```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
np.random.seed(42)
regualar_spenders = np.random.normal(loc=2000, scale= 300, size= 150)
premium_spenders = np.random.normal(loc= 6000, scale= 500, size= 150)
customer_spending = np.concatenate([regualar_spenders, premium_spenders]).reshape(-1,1)
gmm = GaussianMixture(n_components=2, random_state=42)
gmm.fit(customer_spending)
labels= gmm.predict(customer_spending)
plt.figure(figsize=(10,6))
plt.hist(customer_spending[labels==0], bins=30, color='skyblue', alpha=0.7, label='Segment A')
plt.hist(customer_spending[labels==1], bins=30, color='salmon', alpha=0.7, label='Segment B')
plt.title("Customer Segementation using Gaussian Mixture Model")
plt.xlabel("Monthly Spending(INR)")
plt.ylabel("Number of Customers")
plt.legend()
plt.grid(True)
plt.show()
```
